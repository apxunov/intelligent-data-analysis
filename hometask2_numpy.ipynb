# **Задание 1.**
Напишем функцию, возвращающую округленную взвешенную сумму оценок по данным оценкам и весам.
<br> 
В нашем случае вес экзамена равен 0.3, вес домашек - 0.4, вес контрольной - 0.2, вес самостоятельных - 0.1. 
<br>
Например, если за экзамен у вас 7, за домашки 10, за контрольную 8, а за самостоятельные 6, то вы получите отличную оценку 8!

$Mark = 0.3*Ex + 0.4*Hw + 0.2*Cw + 0.1*Test$ 

```
def result_mark(weights: np.array, marks: np.array) -> int:
  return np.dot(weights, marks, out=None).round()
```

```
result_mark([0.3, 0.4, 0.2, 0.1], [7, 10, 8, 6])
```

```
weights = np.array([0.3, 0.4, 0.2, 0.1])
marks = np.array([7, 10, 8, 6])

assert result_mark(weights, marks) == 8
```

```
weights = np.array([0.3, 0.4, 0.2, 0.1])
marks = np.array([7, 0, 8, 6])

assert result_mark(weights, marks) == 4
```

# **Задание 2.**
Напишем функцию, меняющую каждое третье (начиная с 0) значение массива целых чисел на заданное число. 
<br>
Например, если на вход поступает массив `array([3, 5, 1, 0, -3, 22, 213436])` и число `-111`, то на выходе должен получиться массив `array([-111, 5, 1, -111, -3, 22, -111])`.

```
def change_array(array: np.array, number: int) -> np.array:
    array[0::3] = number
    return array
```

```
change_array(np.array([3, 5, 1, 0, -3, 22, 213436]), -111)
```

```
array = np.array([3, 5, 1, 0, -3, 22, 213436])
number = -111

assert np.allclose(change_array(array, number), np.array([-111, 5, 1, -111, -3, 22, -111]))
```

```
array = np.array([3, 14, 15, 92, 6])
number = 8

assert np.allclose(change_array(array, number), np.array([8, 14, 15, 8, 6]))
```

# **Задание 3.**
Напишем функцию, выдающую индексы «близких» элементов заданных массивов, а именно тех пар элементов, чей модуль разницы не превосходит заданного значения. 
<br>
Например, если на вход поступают массив `array([1.5, 0.5, 2, -4.1, -3, 6, -1])`, массив `array([1.2, 0.5, 1, -4, 3, 0, -1.2])` и число `0.5`, то на выходе должен получиться массив `array([0, 1, 3, 6])`
<br> 
**важно:** не `tuple`, а одномерный массив типа `numpy.ndarray` (то есть .`ndim` от него равно 1).

```
def find_close(array1: np.array, array2: np.array, precision: float) -> np.array:
  newArr = np.abs(array1 - array2)
  return np.array(np.where(newArr < precision)).flatten()

find_close(np.array([1.5, 0.5, 2, -4.1, -3, 6, -1]), np.array([1.2, 0.5, 1, -4.0,  3, 0, -1.2]), 0.5)
```

```
array1 = np.array([1.5, 0.5, 2, -4.1, -3, 6, -1])
array2 = np.array([1.2, 0.5, 1, -4.0,  3, 0, -1.2])
precision = 0.5
res = find_close(array1, array2, precision)

assert res.ndim == 1
assert np.allclose(res, np.array([0, 1, 3, 6]))
```

```
array1 = np.array([3.1415, 2.7182, 1.6180, 6.6261])
array2 = np.array([6.6730, 1.3807, -1,     6.0222])
precision = 1.7
res = find_close(array1, array2, precision)

assert res.ndim == 1
assert np.allclose(res, np.array([1, 3]))
```

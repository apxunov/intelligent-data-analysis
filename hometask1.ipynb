```
def getStats(df, colName):
  stats = df[colName].describe()
  stats.name = colName

  min = stats['min']
  max = stats['max']
  meanval = stats['mean']
  std = stats['std']

  statsArr = pd.DataFrame({'Статическая характеристика': ['Минимум', 'Максимум', 'Среднее значение', 'Стандартное отклонение'],
                          'Показатель': [min, max, meanval, std]})

  return statsArr
```

## **Задание 16.**
**Переведем баллы в оценки.** <br>

<u>Система перевода баллов в оценки</u><br>
где $x$ – средний балл студента по трём экзаменам
> $A$ для $x > 90$ <br>
> $B$ для  $x \in [80, 90]$ <br>
> $C$ для  $x \in [70, 80]$ <br>
> $D$ для  $x \in [60, 70]$ <br>
> $E$ для  $x \in [50, 60]$ <br>
> $F$ для  $x < 50 $ <br>

* Создадим вспомогательную функцию, которая будет по среднему баллу за три экзамена выставлять оценку студенту по данным выше критериям.

* Создадим столбец Grade и запишите в него оценку каждого студента.

* Выведем количество студентов, получивших каждую из оценок.

#### 1. Воспользуемся таблицей `subjectsDf`, созданной в 6 задании

```
subjectsDf
```

#### 2. Создадим функицю `mark`, возвращающую оценку, исходя из диапозона, в который попадает средний балл студента по трём экзаменам

```
gradesDict = {90: 'A', 80: 'B', 70: 'C', 60: 'D', 49.999: 'E', -1: 'F'}

def mark(studentsMark):
  for g in gradesDict.keys():
    if studentsMark > g:
      return gradesDict.get(g)
      break
    else:
      continue
  
print({'score>=50': mark(50), 'score<50': mark(49), 'score=60': mark(60), 'score=90': mark(90), 'score>90': mark(95)})
```

#### 3. Добавим в исходную таблицу `df` столбец `Mean score` со средним баллом студента, взятым из столбца `mean` таблицы `subjectDf`.

* Добавим также столбец `Grade`, в который будем выставлять оценку в соответствии с требованиями выше. В этом нам поможет функция `mark`

```
df['Mean score'] = subjectsDf['mean']
df['Grade'] = df['Mean score'].apply(lambda x: mark(x))
df
```
